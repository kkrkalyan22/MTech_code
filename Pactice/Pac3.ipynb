```python
import pandas as pd

# Create dataset
data = {
    'Student_ID': [101, 102, 103, 101, 104],
    'Name': ['Rahul', 'Priya', 'Amit', 'Rahul', 'Ravi'],
    'Study_Hours': [5, 4, 6, 5, 3],
    'Marks': [85, 90, 75, 85, 65]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

print("\nDuplicate Records:")
print(df[df.duplicated()])

df = df.drop_duplicates()

print("\nDataset After Removing Duplicates:")
print(df)
```
