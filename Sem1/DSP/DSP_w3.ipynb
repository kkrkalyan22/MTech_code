```python
import pandas as pd
raw_data = {
    'transaction_id': ['T-101', 'T-102', 'T-103', 'T-104', 'T-105'],
    'date': ['2023-11-01', '2023-11-01', '2023-11-02', '2023-11-02', '2023-11-03'],
    'product': ['Wireless Mouse', 'Mechanical Keyboard', 'Monitor', 'Wireless Mouse', 'Desk Chair'],
    'quantity': [2, 1, 2, 5, 1],
    'unit_price': [25.00, 85.00, 200.00, 25.00, 150.00]
}

df_sales = pd.DataFrame(raw_data)

df_sales['total_sale'] = df_sales['quantity'] * df_sales['unit_price']

print("--- COMPLETE SMART SALES RECORD ---")
print(df_sales)

print("\n--- TOTAL REVENUE BY PRODUCT ---")
revenue_summary = df_sales.groupby('product')['total_sale'].sum().reset_index()

revenue_summary = revenue_summary.sort_values(by='total_sale', ascending=False)
print(revenue_summary.to_string(index=False))

print("\n--- BULK TRANSACTIONS (QTY > 1) ---")
bulk_sales = df_sales[df_sales['quantity'] > 1]
print(bulk_sales[['transaction_id', 'product', 'quantity']])
```
