```python
import numpy as np
marks = [[85,90,80],[70,75,65],[95,92,98],[60,55,70],[88,84,90]]
students = ["Student 1","Student 2","Student 3","Student 4","Student 5"]
subjects = ["Python","Statistics","Data Science"]
total = np.sum(marks, axis = 1)
average = np.mean(marks, axis = 1)
highest = np.max(marks, axis = 0)
lowest = np.min(marks, axis = 0)
print("Students Performance Analysis")
print("--------------------------------------------")
for i in range(len(students)):
    print( students[i], "- Total: ", total[i], "- Average: ",round(average[i],2))
print("\nHighest Marks in Each Subject: ")
for i in range (len(subjects)):
    print(subjects[i],":",highest[i])
print("\nLowest Marks in Each Subject: ")
for i in range (len(subjects)):
    print(subjects[i],":",lowest[i])
print("\nStudents with Average marks >= 75:")
for i in range(len(students)):
    if average[i]>=75:
        print(students[i],"- Average: ",round(average[i],2))
best_index = np.argmax(average)
print("\nBest performance student: ",students[best_index])
print("\nHighest Average: ",round(average[best_index],2))
```
