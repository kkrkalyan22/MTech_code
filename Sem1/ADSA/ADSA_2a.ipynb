```python
#2.a
from collections import deque

graph = {
    'Hostel': ['Canteen', 'Auditorium'],
    'Canteen': ['Hostel', 'Library'],
    'Auditorium': ['Hostel', 'Admin'],
    'Admin': ['Auditorium', 'Library'],
    'Library': ['Canteen', 'Admin']
}

def bfs(graph, start, goal):
    queue = deque([[start]])
    visited = set()

    while queue:
        path = queue.popleft()
        node = path[-1]

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)

            for neighbor in graph[node]:
                if neighbor not in visited:
                    new_path = path + [neighbor]
                    queue.append(new_path)

    return None
path = bfs(graph, 'Hostel', 'Library')

if path:
    print("Shortest path:", " -> ".join(path))
    print("Minimum number of walkways:", len(path) - 1)
else:
    print("No path found!")


```

```python
# Longer 2A program 

from collections import deque

def bfs_print_shortest_path(campus_graph, start_building, target_building):
    # Initialize Queue, Visited set, and a Parent map to track the route
    queue = deque()
    visited = set()
    parent = {}
    
    # Enqueue the starting building
    queue.append(start_building)
    visited.add(start_building)
    parent[start_building] = None  # The start has no parent
    
    path_found = False
    
    # Traverse the campus map
    while queue:
        current_building = queue.popleft()
        
        # Stop searching if we reached the target
        if current_building == target_building:
            path_found = True
            break
            
        # Inspect all walkways extending from the current building
        for neighbor in campus_graph.get(current_building, []):
            if neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = current_building  # Record where we came from
                queue.append(neighbor)
                
    # If the queue emptied without finding the target
    if not path_found:
        print(f"No path exists between {start_building} and {target_building}")
        return -1
        
    # --- Path Reconstruction Phase ---
    path = []
    current = target_building
    
    # Backtrack from the target to the start using the Parent map
    while current is not None:
        path.insert(0, current)  # Prepend to the front of the list
        current = parent[current]
        
    print("Shortest Route: " + " -> ".join(path))
    return len(path) - 1  # Return the number of walkways


# --- Dynamic Input Parsing ---
if __name__ == "__main__":
    # 1. Get number of buildings
    num_buildings = int(input("Enter Number of buildings: "))
    
    # 2. Get space-separated building names
    buildings = input("Enter the names of the Buildings :\n").split()
    
    # 3. Parse the graph connections
    print("Enter the graph :")
    campus_graph = {}
    for i in range(num_buildings):
        line = input().strip()
        # If the line is empty, the building has no outward connections
        if not line:
            neighbors_indices = []
        else:
            neighbors_indices = map(int, line.split())
            
        current_building = buildings[i]
        campus_graph[current_building] = [buildings[idx] for idx in neighbors_indices]
        
    # 4. Get Start and Destination inputs
    start = input("\nStart: ").strip()
    destination = input("Destination : ").strip()
    
    # Run the algorithm
    print("\n--- Output ---")
    num_walkways = bfs_print_shortest_path(campus_graph, start, destination)
    print("Number of Walkways: "num_walkways)
```
