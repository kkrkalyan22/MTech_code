```python
#2A
from pathlib import Path

def list_files_dfs(directory_path: str):
    root = Path(directory_path)
    
    if not root.exists():
        print(f"Error: The path '{directory_path}' does not exist.")
        return
    if not root.is_dir():
        print(f"Error: The path '{directory_path}' is a file, not a directory.")
        return
        
    stack = [root]
    print(f"--- Starting DFS Traversal in: {root.resolve()} ---\n")
    
    while stack:
        current_path = stack.pop()
        
        # 1. Check if it's a directory first
        if current_path.is_dir():
            # Don't print the root itself as a sub-item, only its contents
            if current_path != root:
                print(f"[Dir ] {current_path.relative_to(root)}")
                
            try:
                # Reverse the list to maintain a standard top-to-bottom DFS order
                for child in reversed(list(current_path.iterdir())):
                    stack.append(child)
            except PermissionError:
                print(f"[Access Denied] {current_path.relative_to(root)}")
                
        # 2. If it's not a directory, it's a file
        elif current_path.is_file():
            print(f"[File] {current_path.relative_to(root)}")

target_directory = "/Users/kalyanram/Documents/Coding/python/MTech/Sem1"
list_files_dfs(target_directory)
```

```python
#Shorter 2B program 

#2.b
def dfs(directory):
    for name, content in directory.items():
        if isinstance(content, dict):
            dfs(content)
        else:
            print(name)


directory = {
    "Documents": {
        "adsa notes.txt": None,
        "project.pdf": None
    },
    "pictures": {
        "output.jpg": None
    },
    "sample_code.py": None
}

print("Files inside the directory:")
dfs(directory)
```
