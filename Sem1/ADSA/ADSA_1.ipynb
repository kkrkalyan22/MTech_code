```python
with open(r"/Users/kalyanram/Documents/Coding/python/MTech/Sem1/ADSA/Week1.txt","r") as file:
    T=file.read()
P= input("Enter the search Phrase: ")
def searchPhrase(T,P):
    n=len(T)
    m=len(P)
    for i in range(n-m+1):
        j=0
        while j<m:
            if T[i+j] != P[j]:
                break
            j+=1
        if j==m:
            return i
    return -1
x = searchPhrase(T,P)
if x!= -1:
    print("Search Successful\nThe Phrase is at Index: ",x)
else:
    print("Search Unsuccessful, Phraes is not found")
```
