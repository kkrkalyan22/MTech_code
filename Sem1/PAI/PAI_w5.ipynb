```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create a sample healthcare dataset
data = {
    'Patient_ID': [101, 102, 103, 104, 105,
                   106, 107, 108, 109, 110],

    'Age': [25, 35, 45, 55, 65,
            30, 40, 50, 60, 70],

    'Gender': ['Male', 'Female', 'Male', 'Female',
               'Male', 'Female', 'Male', 'Female',
               'Male', 'Female'],

    'Blood_Pressure': [120, 130, 140, 150, 160,
                       125, 135, 145, 155, 165],

    'Cholesterol': [180, 190, 210, 230, 250,
                    185, 195, 215, 235, 255],

    'Blood_Sugar': [90, 100, 110, 120, 130,
                    95, 105, 115, 125, 135]
}

# Convert data into DataFrame
df = pd.DataFrame(data)

# Display dataset
print("Healthcare Dataset:")
print(df)

# Set visualization style
sns.set_style("whitegrid")

# Create multiple visualizations
plt.figure(figsize=(14, 10))

# 1. Histogram - Age Distribution
plt.subplot(2, 3, 1)
sns.histplot(df['Age'], bins=5, kde=True)
plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Number of Patients")

# 2. Bar Chart - Average Blood Pressure
plt.subplot(2, 3, 2)
sns.barplot(data=df, x='Gender', y='Blood_Pressure')
plt.title("Average Blood Pressure by Gender")
plt.xlabel("Gender")
plt.ylabel("Blood Pressure")

# 3. Scatter Plot - Age vs Cholesterol
plt.subplot(2, 3, 3)
sns.scatterplot(data=df, x='Age',
                y='Cholesterol', hue='Gender')
plt.title("Age vs Cholesterol")
plt.xlabel("Age")
plt.ylabel("Cholesterol")

# 4. Box Plot - Blood Pressure
plt.subplot(2, 3, 4)
sns.boxplot(data=df, x='Gender', y='Blood_Pressure')
plt.title("Blood Pressure Distribution")
plt.xlabel("Gender")
plt.ylabel("Blood Pressure")

# 5. Heatmap - Correlation Analysis
plt.subplot(2, 3, 5)
correlation = df[
    ['Age', 'Blood_Pressure',
     'Cholesterol', 'Blood_Sugar']
].corr()

sns.heatmap(correlation, annot=True,
            cmap='coolwarm', fmt='.2f')

plt.title("Healthcare Data Correlation")

# Adjust layout
plt.tight_layout()

# Display graphs
plt.show()

# Display statistical analysis
print("\nStatistical Analysis:")
print(df.describe())
```
