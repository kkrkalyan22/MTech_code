```python
class Book:
    def __init__(self, book_id, title, author, quantity):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.quantity = quantity
        self.issued_count = 0

    def display(self):
        print(f"ID: {self.book_id} | Title: {self.title} | "
              f"Author: {self.author} | "
              f"Available: {self.quantity - self.issued_count}")

    def issue_book(self):
        if self.quantity > self.issued_count:
            self.issued_count += 1
            return True
        return False

    def return_book(self):
        if self.issued_count > 0:
            self.issued_count -= 1
            return True
        return False


# Parent class
class LibraryMember:
    def __init__(self, member_id, name):
        self.member_id = member_id
        self.name = name

    def get_issue_limit(self):
        return 0

    def display(self):
        print(f"Member ID: {self.member_id}, Name: {self.name}")


# Child class 1
class Student(LibraryMember):
    def get_issue_limit(self):
        return 3


# Child class 2
class Faculty(LibraryMember):
    def get_issue_limit(self):
        return 5


class Library:
    def __init__(self):
        self.books = {}
        self.members = {}
        self.issued_books = {}

    def add_book(self, book):
        self.books[book.book_id] = book
        print("Book added successfully.")

    def register_member(self, member):
        self.members[member.member_id] = member
        self.issued_books[member.member_id] = []
        print("Member registered successfully.")

    def display_books(self):
        print("\n--- College Library Books ---")
        if not self.books:
            print("No books available.")
            return

        for book in self.books.values():
            book.display()

    def issue_book(self, member_id, book_id):
        if member_id not in self.members:
            print("Invalid member ID.")
            return

        if book_id not in self.books:
            print("Invalid book ID.")
            return

        member = self.members[member_id]
        book = self.books[book_id]
        member_books = self.issued_books[member_id]

        if book_id in member_books:
            print("This book is already issued to this member.")
            return

        if len(member_books) >= member.get_issue_limit():
            print("Issue limit reached for this member.")
            return

        if book.issue_book():
            member_books.append(book_id)
            print(f"'{book.title}' issued successfully to {member.name}.")
        else:
            print("Book is currently unavailable.")

    def return_book(self, member_id, book_id):
        if member_id not in self.members:
            print("Invalid member ID.")
            return

        if book_id not in self.books:
            print("Invalid book ID.")
            return

        member_books = self.issued_books[member_id]
        book = self.books[book_id]

        if book_id not in member_books:
            print("This book was not issued to this member.")
            return

        book.return_book()
        member_books.remove(book_id)
        print(f"'{book.title}' returned successfully by "
              f"{self.members[member_id].name}.")

    def display_members(self):
        print("\n--- Registered Library Members ---")
        if not self.members:
            print("No members registered.")
            return

        for member in self.members.values():
            member.display()
            print("Books issued:", self.issued_books[member.member_id])


# Main program
library = Library()

# Add sample books
library.add_book(Book(101, "Python Programming", "Guido van Rossum", 2))
library.add_book(Book(102, "Data Structures", "Mark Allen Weiss", 3))
library.add_book(Book(103, "Machine Learning", "Tom Mitchell", 1))

# Register members
library.register_member(Student(1, "Ravi"))
library.register_member(Faculty(2, "Dr. Kumar"))

while True:
    print("\n===== COLLEGE LIBRARY MANAGEMENT SYSTEM =====")
    print("1. Display Books")
    print("2. Register Student")
    print("3. Register Faculty")
    print("4. Issue Book")
    print("5. Return Book")
    print("6. Display Members")
    print("7. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        library.display_books()

    elif choice == "2":
        member_id = int(input("Enter student ID: "))
        name = input("Enter student name: ")

        if member_id in library.members:
            print("Member ID already exists.")
        else:
            library.register_member(Student(member_id, name))

    elif choice == "3":
        member_id = int(input("Enter faculty ID: "))
        name = input("Enter faculty name: ")

        if member_id in library.members:
            print("Member ID already exists.")
        else:
            library.register_member(Faculty(member_id, name))

    elif choice == "4":
        member_id = int(input("Enter member ID: "))
        book_id = int(input("Enter book ID: "))
        library.issue_book(member_id, book_id)

    elif choice == "5":
        member_id = int(input("Enter member ID: "))
        book_id = int(input("Enter book ID: "))
        library.return_book(member_id, book_id)

    elif choice == "6":
        library.display_members()

    elif choice == "7":
        print("Thank you for using the College Library System.")
        break

    else:
        print("Invalid choice. Please try again.")
```

```python
class Book:
    def __init__(self, book_id, title, author):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.is_issued = False

    def display_details(self):
        print(f"ID: {self.book_id} | Title: {self.title} "
              f"| Author: {self.author}")

    def get_loan_period(self):
        return 14


# Inheritance
class ReferenceBook(Book):
    def get_loan_period(self):
        return 2


class LibraryMember:
    def __init__(self, member_id, name):
        self.member_id = member_id
        self.name = name
        self.borrowed_books = []

    def issue_book(self, book):
        if book.is_issued:
            print(f"Sorry! '{book.title}' is already issued.")
            return

        book.is_issued = True
        self.borrowed_books.append(book)
        print(f"'{book.title}' issued successfully to {self.name}.")
        print(f"Loan period: {book.get_loan_period()} days")

    def return_book(self, book):
        if book not in self.borrowed_books:
            print(f"{self.name} has not borrowed '{book.title}'.")
            return

        book.is_issued = False
        self.borrowed_books.remove(book)
        print(f"'{book.title}' returned successfully by {self.name}.")

    def display_borrowed_books(self):
        print(f"\nBooks borrowed by {self.name}:")
        if not self.borrowed_books:
            print("No books currently borrowed.")
        else:
            for book in self.borrowed_books:
                book.display_details()


class CollegeLibrary:
    def __init__(self):
        self.books = []
        self.members = []

    def add_book(self, book):
        self.books.append(book)
        print(f"Book '{book.title}' added to library.")

    def register_member(self, member):
        self.members.append(member)
        print(f"Member '{member.name}' registered.")

    def display_books(self):
        print("\n--- Library Book Catalogue ---")
        for book in self.books:
            status = "Issued" if book.is_issued else "Available"
            print(f"{book.book_id} | {book.title} | "
                  f"{book.author} | {status}")

    def find_book(self, book_id):
        for book in self.books:
            if book.book_id == book_id:
                return book
        return None


# Main program
library = CollegeLibrary()

book1 = Book(101, "Python Programming", "Guido van Rossum")
book2 = Book(102, "Data Structures", "Mark Allen Weiss")
book3 = ReferenceBook(103, "Machine Learning", "Tom Mitchell")

member1 = LibraryMember(1, "Sandeep")

library.add_book(book1)
library.add_book(book2)
library.add_book(book3)
library.register_member(member1)

while True:
    print("\n===== COLLEGE LIBRARY MANAGEMENT =====")
    print("1. Display all books")
    print("2. Issue a book")
    print("3. Return a book")
    print("4. Display borrowed books")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        library.display_books()

    elif choice == "2":
        book_id = int(input("Enter book ID to issue: "))
        book = library.find_book(book_id)

        if book:
            member1.issue_book(book)
        else:
            print("Book not found.")

    elif choice == "3":
        book_id = int(input("Enter book ID to return: "))
        book = library.find_book(book_id)

        if book:
            member1.return_book(book)
        else:
            print("Book not found.")

    elif choice == "4":
        member1.display_borrowed_books()

    elif choice == "5":
        print("Thank you for using the library system!")
        break

    else:
        print("Invalid choice. Please try again.")
```
