```python
import numpy as np
a = np.arange(1,5).reshape(2,2)
b = np.arange(5,9).reshape(2,2)
print("Matrix A:")
print(a)
print("Matrix b:")
print(b)
print("1. Addition :\n", a+b)
print("2. Multiplication :\n", a@b)
print("3. Transpose of A:\n", a.T)
print("4. Inverse of A:")
try:
    inverse_A = np.linalg.inv(a)
    print(inverse_A)
except np.linalg.LinAlgError:
    print("Matrix A is singular; inverse does not exist.")
if np.linalg.det(a) != 0:
    print("\nVerification (A @ inverse_A):")
    print(np.round(a @ inverse_A, 2))
```
