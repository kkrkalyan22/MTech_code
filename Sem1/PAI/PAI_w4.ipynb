```python
import pandas as pd

# Step 1: Create a retail sales dataset
data = {
    'Order_ID': [101, 102, 103, 104, 105, 103],
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Laptop', None, 'Tablet'],
    'Quantity': [2, 5, 3, None, 4, 3],
    'Unit_Price': [50000, 15000, 20000, 50000,None, 20000],
    'Sales_Date': ['2026-01-01', '2026-01-02','2026-01-03', '2026-01-04','2026-01-05', '2026-01-03']
}

df = pd.DataFrame(data)

print("Original Retail Sales Dataset:")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())

df = df.drop_duplicates()

df['Product'] = df['Product'].fillna('Unknown')
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].mean())
df['Unit_Price'] = df['Unit_Price'].fillna(df['Unit_Price'].mean())

df['Sales_Date'] = pd.to_datetime(df['Sales_Date'])

df['Total_Sales'] = df['Quantity'] * df['Unit_Price']

print("\nCleaned Retail Sales Dataset:")
print(df.round(2))

total_revenue = df['Total_Sales'].sum()

print("\nTotal Revenue:", round(total_revenue, 2))

product_sales = df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False)

print("\nProduct-wise Revenue:")
print(product_sales.round(2))

best_product = product_sales.idxmax()

print("\nBest-performing Product:", best_product)

print("\nSales Summary:")
print(df[['Quantity', 'Unit_Price','Total_Sales']].describe().round(2))
```
