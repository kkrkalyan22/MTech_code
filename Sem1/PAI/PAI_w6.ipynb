```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder, LabelEncoder
data = {
   'Department': ['CSE', 'ece', 'AIML', ' CSE ', 'ECE', 'aiml'],
   'Study_Hours': [5, 3, 8, 2, None, 4],
   'Attendance': [85, 70, 95, None, 80, 75],
   'Lab_Performance': [80, 65, 90, 50, 75, 70],
   'Result': ['Pass', 'Pass', 'Pass', 'Fail', 'Pass', 'Fail']
}
df = pd.DataFrame(data)
print("Original Dataset:")
print(df)
df['Department'] = (
   df['Department'].str.strip().str.upper()
)
num_cols = ['Study_Hours', 'Attendance', 'Lab_Performance']
imputer = SimpleImputer(strategy='mean')
df[num_cols] = imputer.fit_transform(df[num_cols])
print("\nAfter Handling Missing Values:")
print(df)
scaler = MinMaxScaler()
df[num_cols] = scaler.fit_transform(df[num_cols])
print("\nAfter Feature Scaling:")
print(df)
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
encoded = encoder.fit_transform(df[['Department']])
encoded_df = pd.DataFrame(
   encoded,
   columns=encoder.get_feature_names_out(['Department']),
   index=df.index
)
label_encoder = LabelEncoder()
df['Result'] = label_encoder.fit_transform(df['Result'])
df = pd.concat(
   [df.drop(columns=['Department']), encoded_df],
   axis=1
)
print("\nFinal Preprocessed Dataset:")
print(df)
print("\nDataset is ready for Machine Learning!")
```
